```python
def inspect(dia: float) -> str:
    return "UNDER" if dia < 9.8 else "OVER" if dia > 10.2 else "PASS"

def tally(labels: list[str]) -> dict[str, int]:
    return {k: labels.count(k) for k in ("PASS", "UNDER", "OVER")}

def report(shift: str, counts: dict[str, int]) -> str:
    return f"[{shift.upper()}] PASS: {counts.get('PASS', 0)} | UNDER: {counts.get('UNDER', 0)} | OVER: {counts.get('OVER', 0)}"

# --- Tests ---
print(inspect(9.5), inspect(10.0), inspect(10.5))  # UNDER PASS OVER
sample_counts = tally(["PASS", "PASS", "UNDER", "OVER"])
print(sample_counts)                                # {'PASS': 2, 'UNDER': 1, 'OVER': 1}
print(report("Morning", sample_counts))             # [MORNING] PASS: 2 | UNDER: 1 | OVER: 1
```

```python
import lab04_autobench as old

def inspect(dia: float) -> str:
    return "UNDER" if dia < 9.8 else "OVER" if dia > 10.2 else "PASS"

def tally(labels: list[str]) -> dict[str, int]:
    return {k: labels.count(k) for k in ("PASS", "UNDER", "OVER")}

def report(shift: str, counts: dict[str, int]) -> str:
    return f"[{shift.upper()}] PASS: {counts.get('PASS', 0)} | UNDER: {counts.get('UNDER', 0)} | OVER: {counts.get('OVER', 0)}"

# Refactored pipeline using the 3 jigs
def run_bench(shift: str, readings: list[float]) -> str:
    return report(shift, tally([inspect(d) for d in readings]))

# Verification & Line Count
readings = [10.0, 9.5, 10.5, 10.1, 9.7]
print(f"Old Output: {old.run_bench('Morning', readings)}")
print(f"New Output: {run_bench('Morning', readings)}")

# Record Line Counts
# Before (Old lab04_autobench.py): __ lines
# After  (Refactored version)    : __ lines
# Difference                     : __ lines reduced
```

```python
def lookup(catalogue: dict, code: str, default: float = 0.0) -> float:
    return catalogue.get(code, default)

def scrap_codes(catalogue: dict) -> set[str]:
    return {code for code, dia in catalogue.items() if dia < 9.8 or dia > 10.2}

# --- Tests ---
cat = {"P01": 10.0, "P02": 9.5, "P03": 10.5, "P04": 10.1}

print(lookup(cat, "P01"))      # 10.0
print(lookup(cat, "P99"))      # 0.0 (safe fallback)
print(scrap_codes(cat))        # {'P02', 'P03'}
```

```python
def inspect(d: float) -> str: return "UNDER" if d < 9.8 else "OVER" if d > 10.2 else "PASS"
def tally(lbls: list) -> dict: return {k: lbls.count(k) for k in ("PASS", "UNDER", "OVER")}
def report(shift: str, c: dict) -> str: return f"[{shift.upper()}] PASS: {c['PASS']} | UNDER: {c['UNDER']} | OVER: {c['OVER']}"
def lookup(cat: dict, code: str, default: float = 0.0) -> float: return cat.get(code, default)
def scrap_codes(cat: dict) -> set: return {code for code, d in cat.items() if inspect(d) != "PASS"}
```
