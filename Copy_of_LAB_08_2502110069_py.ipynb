```python
# lab05.py

def parse(data):
    report, rejects = [], []
    for line in data:
        try:
            if not isinstance(line, str) or not line.strip():
                raise ValueError("Empty or invalid input")
            k, v = line.split(",")
            report.append((k.strip(), float(v)))
        except (ValueError, TypeError, AttributeError) as e:
            rejects.append((line, f"{type(e).__name__}: {e}"))

    vals = [v for _, v in report]
    avg = sum(vals) / len(vals) if vals else 0.0
    return report, rejects, avg

report, rejects, avg = parse(["a,10", "", "  ", None, "b,bad", "c,20,extra"])

print(f"REPORT: {report}\nAVERAGE: {avg}\nREJECTS:")
for line, reason in rejects:
    print(f"{repr(line)} -> {reason}")
```

```python
class SpecViolation(Exception):
    """Raised when a work order job violates specifications."""
    pass

class Machine:
    def process(self, job):
        if not isinstance(job, dict) or job.get("status") == "bad":
            raise SpecViolation(f"Invalid job spec: {job}")
        return f"Processed {job['id']}"

# Fleet execution
fleet = Machine()
jobs = [{"id": 101, "status": "ok"}, {"id": 102, "status": "bad"}, "malformed", {"id": 103, "status": "ok"}]

results, rejects = [], []

for job in jobs:
    try:
        results.append(fleet.process(job))
    except SpecViolation as e:
        rejects.append((job, str(e)))

print(f"RESULTS: {results}")
print("REJECTS:")
for j, r in rejects:
    print(f"{repr(j)} -> {r}")
```
