IT-10

```python
import random

# Initialize counters for the shift report
accept_count = 0
rework_count = 0
scrap_count = 0

# Run 100 parts through the bench
for i in range(1, 101):
    # Generate diameter rounded to 2 decimal places
    diameter = round(random.uniform(24.90, 25.10), 2)

    # Job 09 three-bin sorting ladder
    if 24.95 <= diameter <= 25.05:
        bin_status = "ACCEPT"
        accept_count += 1
    elif diameter > 25.05:
        bin_status = "REWORK"
        rework_count += 1
    else:  # diameter < 24.95
        bin_status = "SCRAP"
        scrap_count += 1

    # Announce each part
    print(f"Part {i:03d}: {diameter:.2f} mm | Bin: {bin_status}")

# Shift Report
print("\n--- SHIFT REPORT ---")
print(f"ACCEPT: {accept_count}")
print(f"REWORK: {rework_count}")
print(f"SCRAP:  {scrap_count}")
print(f"TOTAL:  {accept_count + rework_count + scrap_count}")
```

IT-8

```python
import random

# Operator Prompt: Select operating mode
print("--- BENCH CONFIGURATION ---")
mode = input("Select mode (1 = Halt on Scrap, 2 = Skip / Continue): ").strip()

# Initialize counters
accept_count = 0
rework_count = 0
scrap_count = 0
inspected_count = 0

print("\n--- STARTING PRODUCTION RUN ---")

for i in range(1, 101):
    inspected_count += 1
    diameter = round(random.uniform(24.90, 25.10), 2)

    # Job 09 Sorting Ladder
    if 24.95 <= diameter <= 25.05:
        bin_status = "ACCEPT"
        accept_count += 1
    elif diameter > 25.05:
        bin_status = "REWORK"
        rework_count += 1
    else:  # diameter < 24.95 (SCRAP)
        bin_status = "SCRAP"
        scrap_count += 1

    # Announce the current part
    print(f"Part {i:03d}: {diameter:.2f} mm | Bin: {bin_status}")

    # Scrap Handling Logic
    if bin_status == "SCRAP":
        if mode == "1":  # Halt Mode
            print(f"\n⚠️ CRITICAL: SCRAP detected! Halting production line.")
            break
        elif mode == "2":  # Skip Mode
            print("⚠️ SCRAP logged. Continuing run...")
            continue

# Shift Report
print("\n" + "=" * 30)
print("       SHIFT REPORT      ")
print("=" * 30)
print(f"Operating Mode  : {'Halt on Scrap' if mode == '1' else 'Skip & Continue'}")
print(f"Parts Inspected : {inspected_count}")
print(f"ACCEPT          : {accept_count}")
print(f"REWORK          : {rework_count}")
print(f"SCRAP           : {scrap_count}")
print("=" * 30)
```

```python
import random

consecutive_accepts = 0
total_parts = 0
SAFETY_CAP = 500

# Keep machining until 5 ACCEPTs in a row or safety cap hit
while consecutive_accepts < 5 and total_parts < SAFETY_CAP:
    total_parts += 1
    diameter = round(random.uniform(24.90, 25.10), 2)

    # Job 09 Sorting Ladder
    if 24.95 <= diameter <= 25.05:
        bin_status = "ACCEPT"
        consecutive_accepts += 1
    elif diameter > 25.05:
        bin_status = "REWORK"
        consecutive_accepts = 0  # Reset streak
    else:
        bin_status = "SCRAP"
        consecutive_accepts = 0  # Reset streak

    print(
        f"Part {total_parts:03d}: {diameter:.2f} mm | Bin: {bin_status:<6} | Streak: {consecutive_accepts}"
    )

# Production Summary
print("\n" + "=" * 35)
if consecutive_accepts == 5:
    print(f" SUCCESS: Target achieved in {total_parts} parts!")
else:
    print(f"⚠️ SAFETY HALT: Hit maximum limit of {SAFETY_CAP} parts.")
print("=" * 35)
```
