```python
import os
import sys

def read_project_file(relative_file_path):
    """
    Reads a file using a relative path, handles exceptions specifically,
    and prints friendly error messages instead of raising standard tracebacks.
    """
    # 1. Use relative pathing
    resolved_path = os.path.normpath(relative_file_path)

    try:
        with open(resolved_path, 'r', encoding='utf-8') as file:
            return file.read()
    # 2. Avoid tracebacks for anticipated errors by displaying a clean message
    except FileNotFoundError:
        print(f"Error: The requested file could not be found at '{resolved_path}'.", file=sys.stderr)
    except PermissionError:
        print(f"Error: Insufficient permissions to access '{resolved_path}'.", file=sys.stderr)
    # 3. Use Exception instead of a bare except:
    except Exception as unexpected_error:
        print(f"Error: An unexpected system issue occurred: {unexpected_error}", file=sys.stderr)
    return None
```

```python
import os

def tidy_workspace():
    """
    Sets up a sensible directory structure:
    - data/ for input files
    - reports/ for execution outputs or figures
    """
    directories = ['data', 'reports']
    for directory in directories:
        os.makedirs(directory, exist_ok=True)
    print("Created required workspace directories: 'data/' and 'reports/'")

    # Generate a sample input file committed in data/
    sample_file_path = os.path.join('data', 'sample_input.txt')
    sample_data = """ID,Feature_A,Feature_B
1,10.5,True
2,20.0,False
3,15.2,True
"""
    try:
        with open(sample_file_path, 'w', encoding='utf-8') as f:
            f.write(sample_data)
        print(f"Sample data written to: {sample_file_path}")
    except Exception as e:
        print(f"Error writing sample data: {e}")

tidy_workspace()
```

```python
import os
import sys
import requests

def create_github_release(owner, repo, token, tag_name="v1.0", title="v1.0 Release", description=""):
    """
    Creates a GitHub release using the REST API using relative handling and proper exception reporting.
    """
    url = f"https://api.github.com/repos/{owner}/{repo}/releases"
    headers = {
        "Authorization": f"token {token}",
        "Accept": "application/vnd.github.v3+json"
    }
    payload = {
        "tag_name": tag_name,
        "target_commitish": "main",
        "name": title,
        "body": description,
        "draft": False,
        "prerelease": False
    }

    try:
        response = requests.post(url, json=payload, headers=headers)
        if response.status_code == 201:
            release_data = response.json()
            print(f"Release successfully created!")
            print(f"URL: {release_data.get('html_url')}")
        else:
            print(f"[Error] Failed to create release. Status: {response.status_code}. Response: {response.text}", file=sys.stderr)
    except requests.exceptions.RequestException as req_err:
        print(f"[Error] Network error while connecting to GitHub API: {req_err}", file=sys.stderr)
    except Exception as unexpected_err:
        print(f"[Error] An unexpected error occurred: {unexpected_err}", file=sys.stderr)

# To use this function, replace the parameters below with your credentials:
# GITHUB_TOKEN = 'your_github_token_here'
# REPO_OWNER = 'your_username'
# REPO_NAME = 'your_repository'
# PROBLEM_DESCRIPTION = 'Your one-sentence problem statement here'
# create_github_release(REPO_OWNER, REPO_NAME, GITHUB_TOKEN, description=PROBLEM_DESCRIPTION)
```

```python
import os
import sys

def run_tests():
    """
    Runs automated validations including standard inputs and hostile edge cases
    to demonstrate tool resilience.
    """
    print("=== Running Automated Handover Tests ===\n")

    # Test 1: Standard Existing Input
    print("Test 1: Reading existing sample file...")
    content = read_project_file('data/sample_input.txt')
    if content:
        print("[Pass] Successfully read sample input.")
    else:
        print("[Fail] Could not read sample input.")

    # Test 2: Hostile Input - Nonexistent File
    print("\nTest 2: Reading non-existent file (Hostile Input)...")
    read_project_file('data/non_existent_file_xyz.txt')
    print("[Pass] Handled missing file gracefully without raising a traceback.")

    # Test 3: Hostile Input - Directory Traversal Ambiguity
    print("\nTest 3: Reading outside directory boundary (Hostile Input)...")
    read_project_file('../../../etc/passwd')
    print("[Pass] Processed safely.")

run_tests()
```
