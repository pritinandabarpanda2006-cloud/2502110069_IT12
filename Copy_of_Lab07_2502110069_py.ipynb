```python
class Machine:
    def __init__(self, name: str, limit_rpm: int):
        self.name = name
        self.limit_rpm = limit_rpm
        self.active = False

    def __str__(self):
        return f"{self.name} (Limit: {self.limit_rpm} RPM) - Active: {self.active}"

    def start(self):
        self.active = True


# Commission 2 machines
m1 = Machine("Lathe", 3000)
m2 = Machine("Mill", 5000)

# Modify m1 only
m1.start()

# Prove independence: m1 is Active (True), m2 stays inactive (False)
print(m1)
print(m2)
```

```python
class Machine:
    def __init__(self, name: str, max_rpm: int):
        self.name = name
        self.max_rpm = max_rpm
        self.__speed = 0  # Private state

    def set_speed(self, rpm: int):
        """Validating operating method with boundary guard."""
        if 0 <= rpm <= self.max_rpm:
            self.__speed = rpm
            print(f"{self.name} speed set to {self.__speed} RPM.")
        else:
            print(f"Error: {rpm} RPM is out of bounds (0-{self.max_rpm}). Command refused.")

    def status(self) -> str:
        """Status getter."""
        return f"{self.name} | Current Speed: {self.__speed}/{self.max_rpm} RPM"


# Commission machine
lathe = Machine("Lathe", 3000)

# Valid command
lathe.set_speed(1500)

# Out-of-range command (Refused by guard)
lathe.set_speed(4000)

# Check state via getter
print(lathe.status())
```

```python
import random

class Machine:
    def __init__(self, name: str):
        self.name = name
        self.__processed = 0
        self.__failed = 0

    def process_job(self, units: int):
        if units > 0:
            self.__processed += units
        else:
            self.__failed += 1

    def status(self) -> str:
        return f"[{self.name} Report] Units Processed: {self.__processed} | Failed Jobs: {self.__failed}"

# Initialize machine and run a batch of 10 jobs
m = Machine("Lathe-01")
batch = [random.randint(-2, 10) for _ in range(10)]  # Negative values represent invalid jobs

for job in batch:
    m.process_job(job)

print(m.status())
```
