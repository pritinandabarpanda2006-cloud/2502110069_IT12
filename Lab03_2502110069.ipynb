```python
# batch_inspector.py
# Three-bin quality sort for 10 manufactured parts

accept_count = 0
rework_count = 0
scrap_count = 0

target_diameter = 50.0

print("--- Batch Inspector: Live Inspection ---")
for i in range(1, 11):
    diameter = float(input(f"Enter diameter for part {i} (mm): "))

    # Three-bin sort logic based on deviation from target
    deviation = abs(diameter - target_diameter)

    if deviation <= 0.05:
        print("  -> ACCEPT")
        accept_count += 1
    elif deviation <= 0.15:
        print("  -> REWORK")
        rework_count += 1
    else:
        print("  -> SCRAP")
        scrap_count += 1

# Final report line
print("\n--- Final Batch Report ---")
print(f"Inspected: 10 parts | ACCEPT: {accept_count} | REWORK: {rework_count} | SCRAP: {scrap_count}")
```

```python
def audit_parts(total, size):
    boxes, loose = total // size, total % size
    print(f"Boxes: {boxes} | Loose: {loose}")
    print(f"Audit: {boxes} * {size} + {loose} == {total} -> {boxes * size + loose == total}")

# Example
audit_parts(145, 12)
```

```python
def shift_audit(batch, box_size=4):
    bins = {"Accepted": batch.count("pass"), "Rejected": batch.count("fail"), "Rework": batch.count("rework")}
    boxes, loose = divmod(bins["Accepted"], box_size)
    print(f"Bins: {bins}\nBoxes: {boxes} | Loose: {loose}\nAudit: {boxes} * {box_size} + {loose} == {bins['Accepted']} -> {boxes * box_size + loose == bins['Accepted']}")

# Example
shift_audit(["pass", "pass", "fail", "pass", "rework", "pass", "pass", "pass"])
```
