```python
# A list of 10 sample log lines with messy spacing
log_lines = [
    "A101    12.3   0.50",
    "B202  15.0  0.60",
    "C303     11.8    0.55",
    "D404 14.5  0.70",
    "E505    13.2   0.65",
    "F606  12.3  0.50", # Duplicate diameter
    "G707     16.1    0.80",
    "H808 10.9  0.45",
    "I909    13.7   0.72",
    "J010  14.0  0.68"
]

diameters = []

for line in log_lines:
    # Split by whitespace, which handles messy spacing
    parts = line.split()
    try:
        # The diameter is the second part (index 1)
        diameter = float(parts[1])
        diameters.append(diameter)
    except (IndexError, ValueError) as e:
        print(f"Skipping malformed line: {line} - Error: {e}")

# Report count, average, min, max
count = len(diameters)
average = sum(diameters) / count if count > 0 else 0
minimum = min(diameters) if count > 0 else None
maximum = max(diameters) if count > 0 else None

print(f"\n--- Diameter Statistics ---")
print(f"Count: {count}")
print(f"Average: {average:.2f}")
print(f"Minimum: {minimum:.2f}")
print(f"Maximum: {maximum:.2f}")
```

```python
# Define the thresholds for the three-bin sort
# These values are illustrative and can be adjusted based on specifications
ACCEPTED_MIN = 12.0
ACCEPTED_MAX = 14.5
REWORK_MIN_LOW = 11.5
REWORK_MAX_HIGH = 15.0

# Initialize lists for each bin
accepted_diameters = []
rework_diameters = []
scrap_diameters = []
scrapped_part_codes = []

# Iterate through the original log lines to perform the sort
for line in log_lines:
    parts = line.split()
    try:
        part_code = parts[0]
        diameter = float(parts[1])

        if ACCEPTED_MIN <= diameter <= ACCEPTED_MAX:
            accepted_diameters.append(diameter)
        elif (REWORK_MIN_LOW <= diameter < ACCEPTED_MIN) or (ACCEPTED_MAX < diameter <= REWORK_MAX_HIGH):
            rework_diameters.append(diameter)
        else:
            scrap_diameters.append(diameter)
            scrapped_part_codes.append(part_code)
    except (IndexError, ValueError) as e:
        print(f"Skipping malformed line during sorting: {line} - Error: {e}")

# Report counts for each bin
print(f"\n--- Three-Bin Sort Results ---")
print(f"Accepted Count: {len(accepted_diameters)}")
print(f"Rework Count: {len(rework_diameters)}")
print(f"Scrap Count: {len(scrap_diameters)}")

# Report accepted average
accepted_average = sum(accepted_diameters) / len(accepted_diameters) if accepted_diameters else 0
print(f"Accepted Average Diameter: {accepted_average:.2f}")

# List scrapped part codes
print(f"Scrapped Part Codes: {scrapped_part_codes}")
```

```python
# Define separate log lines for Day and Night shifts
# Note: I'll create new log lines for demonstration. In a real scenario, these would come from distinct sources.
# Some overlaps and unique items are included for set comparison later.

day_shift_log_lines = [
    "DA101    12.3   0.50",
    "DB202  15.0  0.60", # Rework
    "DC303     11.8    0.55", # Rework
    "DD404 14.5  0.70",
    "DE505    13.2   0.65",
    "DF606  17.0  0.50", # Scrap
    "DG707     11.0    0.80", # Scrap
    "DH808 13.5  0.45",
    "DI909    14.0   0.72",
    "DJ010  12.5  0.68"
]

night_shift_log_lines = [
    "NA111    12.2   0.51",
    "NB222  15.1  0.61", # Rework
    "NC333     11.9    0.56", # Rework
    "ND444 14.4  0.71",
    "NE555    13.1   0.66",
    "DG707  10.0  0.80", # Scrap (common with day shift)
    "NF777     16.5    0.81", # Scrap
    "NG888 13.6  0.46",
    "NH999    14.1   0.73",
    "DA101  12.3  0.69" # Accepted (common with day shift)
]

# --- Three-bin sort function for reusability ---
def perform_three_bin_sort(log_lines, ACCEPTED_MIN, ACCEPTED_MAX, REWORK_MIN_LOW, REWORK_MAX_HIGH):
    accepted = []
    rework = []
    scrap_part_codes = set() # Using a set for part codes to easily find unique/common later

    for line in log_lines:
        parts = line.split()
        try:
            part_code = parts[0]
            diameter = float(parts[1])

            if ACCEPTED_MIN <= diameter <= ACCEPTED_MAX:
                accepted.append(diameter)
            elif (REWORK_MIN_LOW <= diameter < ACCEPTED_MIN) or (ACCEPTED_MAX < diameter <= REWORK_MAX_HIGH):
                rework.append(diameter)
            else:
                scrap_part_codes.add(part_code)
        except (IndexError, ValueError) as e:
            print(f"Skipping malformed line during sorting: {line} - Error: {e}")
    return accepted, rework, scrap_part_codes

# --- Process Day Shift ---
print("\n--- Day Shift Report ---")
day_accepted_diameters, day_rework_diameters, day_scrapped_part_codes = perform_three_bin_sort(
    day_shift_log_lines, ACCEPTED_MIN, ACCEPTED_MAX, REWORK_MIN_LOW, REWORK_MAX_HIGH
)

day_total_parts = len(day_shift_log_lines)
day_accepted_count = len(day_accepted_diameters)
day_accept_rate = (day_accepted_count / day_total_parts) * 100 if day_total_parts > 0 else 0
print(f"Day Shift Accept Rate: {day_accept_rate:.2f}%")
print(f"Day Shift Scrapped Part Codes: {day_scrapped_part_codes}")

# --- Process Night Shift ---
print("\n--- Night Shift Report ---")
night_accepted_diameters, night_rework_diameters, night_scrapped_part_codes = perform_three_bin_sort(
    night_shift_log_lines, ACCEPTED_MIN, ACCEPTED_MAX, REWORK_MIN_LOW, REWORK_MAX_HIGH
)

night_total_parts = len(night_shift_log_lines)
night_accepted_count = len(night_accepted_diameters)
night_accept_rate = (night_accepted_count / night_total_parts) * 100 if night_total_parts > 0 else 0
print(f"Night Shift Accept Rate: {night_accept_rate:.2f}%")
print(f"Night Shift Scrapped Part Codes: {night_scrapped_part_codes}")

# --- Compare Scrapped Part Codes using Sets ---
print("\n--- Scrapped Part Code Comparison ---")

# Part codes appearing in both shifts' scrap lists
common_scrapped_codes = day_scrapped_part_codes.intersection(night_scrapped_part_codes)
print(f"Scrapped in Both Shifts: {common_scrapped_codes}")

# Part codes unique to Day Shift scrap list
unique_to_day_scrap = day_scrapped_part_codes.difference(night_scrapped_part_codes)
print(f"Unique to Day Shift Scrap: {unique_to_day_scrap}")

# Part codes unique to Night Shift scrap list
unique_to_night_scrap = night_scrapped_part_codes.difference(day_scrapped_part_codes)
print(f"Unique to Night Shift Scrap: {unique_to_night_scrap}")
```
